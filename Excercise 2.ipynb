```python
# Author: Taha Basharat
# Date: October 21, 2025
# This program keeps only two dinner guests because the new table didn’t arrive.

people = ["Hamza", "Ali", "Ahsan", "Umer", "Jawad", "Hassan"]

print("Unfortunately, the new dinner table won't arrive on time, and I can invite only two people.\n")

# Remove guests until only two remain
while len(people) > 2:
    removed_guest = people.pop()
    print(f"Sorry {removed_guest}, I can't invite you to dinner this time.")

print("\nInvitations to remaining guests:")
for p in people:
    print(f"{p}, you're still invited to dinner!")

# Remove the last two names
del people[0]
del people[0]

# Verify that the list is empty
print("\nFinal guest list:", people)
```

```python
# Author: Taha Basharat
# Date: October 21, 2025
# This program stores places you'd like to visit and demonstrates list sorting methods.

places = ["Japan", "Switzerland", "Turkey", "Canada", "Maldives"]

print("Original order:", places)
print("Alphabetical order:", sorted(places))
print("Still original order:", places)
print("Reverse alphabetical order:", sorted(places, reverse=True))
print("Still original order:", places)

# Reverse the list
places.reverse()
print("Reversed order:", places)

# Reverse it back to original
places.reverse()
print("Back to original order:", places)

# Sort permanently in alphabetical order
places.sort()
print("Alphabetical order (permanent):", places)

# Sort permanently in reverse alphabetical order
places.sort(reverse=True)
print("Reverse alphabetical order (permanent):", places)
```

```python
# Author: Taha Basharat
# Date: October 21, 2025
# This program performs a variety of conditional tests.

car = "subaru"

print("Is car == 'subaru'? I predict True.")
print(car == "subaru")

print("\nIs car == 'audi'? I predict False.")
print(car == "audi")

print("\nIs car != 'toyota'? I predict True.")
print(car != "toyota")

print("\nIs car.lower() == 'subaru'? I predict True.")
print(car.lower() == "subaru")

print("\nIs car.upper() == 'SUBARU'? I predict True.")
print(car.upper() == "SUBARU")

# Numerical tests
age = 20
print("\nIs age > 18? I predict True.")
print(age > 18)

print("\nIs age < 18? I predict False.")
print(age < 18)

# Using and/or
print("\nIs age > 18 and age < 30? I predict True.")
print(age > 18 and age < 30)

print("\nIs age < 18 or age > 25? I predict False.")
print(age < 18 or age > 25)

# List tests
cars = ["subaru", "audi", "bmw"]
print("\nIs 'bmw' in cars? I predict True.")
print("bmw" in cars)

print("\nIs 'toyota' not in cars? I predict True.")
print("toyota" not in cars)
```

```python
# Author: Taha Basharat
# Date: October 21, 2025
# This program demonstrates more types of conditional comparisons.

fruit = "Apple"

# Equality and inequality with strings
print(fruit == "Apple")  # True
print(fruit != "Banana")  # True

# Using lower()
print(fruit.lower() == "apple")  # True
print(fruit.lower() == "APPLE")  # False

# Numerical comparisons
num = 10
print(num == 10)        # True
print(num != 5)         # True
print(num > 5)          # True
print(num < 20)         # True
print(num >= 10)        # True
print(num <= 9)         # False

# Using and/or
print(num > 5 and num < 15)  # True
print(num < 5 or num == 10)  # True

# In and not in
colors = ["red", "green", "blue"]
print("red" in colors)       # True
print("yellow" not in colors)  # True
```

```python
# Ensure new usernames are unique (case-insensitive)
current_users = ["Ali", "Sara", "John", "Fatima", "David"]
new_users = ["john", "Bilal", "ALI", "Mary", "Taha"]

# Convert current users to lowercase for comparison
current_users_lower = [u.lower() for u in current_users]

for new_user in new_users:
    if new_user.lower() in current_users_lower:
        print(f"Sorry, the username '{new_user}' is already taken.")
    else:
        print(f"The username '{new_user}' is available.")
```

```python
# Display ordinal numbers (1st, 2nd, 3rd, etc.)
numbers = list(range(1, 10))

for num in numbers:
    if num == 1:
        print("1st")
    elif num == 2:
        print("2nd")
    elif num == 3:
        print("3rd")
    else:
        print(f"{num}th")
```
